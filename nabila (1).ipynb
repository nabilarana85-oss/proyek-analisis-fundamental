## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Bagaimana perbandingan rata-rata jumlah penyewaan sepeda per jam antara hari kerja dan hari libur (bukan hari kerja) sepanjang tahun 2011–2012, dan pada jam berapa masing-masing jenis hari mencapai rata-rata penyewaan tertinggi, sehingga pengelola dapat menentukan jam yang perlu ditambah ketersediaan sepedanya?

- **Pertanyaan 2:** Bagaimana rata-rata jumlah penyewaan sepeda harian berdasarkan kombinasi musim dan kondisi cuaca sepanjang tahun 2011–2012, dan kombinasi musim-cuaca mana yang menghasilkan rata-rata tertinggi serta terendah, sehingga pengelola dapat menyesuaikan jumlah armada dan strategi operasional?

#### Memeriksa Kualitas Data (Missing Value, Duplikat, dan Outlier)

```python
# Memeriksa informasi dataset
print("Informasi Hour Dataset:")
hour_df.info()

print("\nInformasi Day Dataset:")
day_df.info()

# Memeriksa missing values
print("\nMissing Values Hour:")
display(hour_df.isnull().sum())

print("\nMissing Values Day:")
display(day_df.isnull().sum())

# Memeriksa data duplikat
print("\nJumlah Duplikat Hour:", hour_df.duplicated().sum())
print("Jumlah Duplikat Day:", day_df.duplicated().sum())
```

```python
print("Statistik Hour Dataset:")
display(hour_df[['temp', 'hum', 'windspeed', 'cnt']].describe())

print("Statistik Day Dataset:")
display(day_df[['temp', 'hum', 'windspeed', 'cnt']].describe())
```

```python
columns = ['temp', 'hum', 'windspeed', 'cnt']

for name, df in [('Hour', hour_df), ('Day', day_df)]:
    print(f"\nOutlier pada {name} Dataset:")
    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = df[(df[col] < lower) | (df[col] > upper)]

        print(f"{col}: {len(outliers)} outlier")
```

**Steps to Take:**
1. Memeriksa struktur dan tipe data menggunakan `info()` untuk mengetahui kesesuaian tipe data setiap kolom.
2. Memeriksa missing value menggunakan `isna().sum()` untuk mengetahui apakah terdapat data kosong.
3. Memeriksa data duplikat menggunakan `duplicated().sum()` untuk mengetahui apakah terdapat baris yang tercatat lebih dari satu kali.
4. Memeriksa statistik deskriptif menggunakan `describe()` untuk memahami distribusi data dan mendeteksi kemungkinan nilai yang tidak wajar.
5. Menindaklanjuti hasil pemeriksaan dengan mengubah kolom `dteday` menjadi tipe `datetime` serta mengubah kode kategori menjadi label yang mudah dipahami pada tahap Cleaning Data.

### Cleaning Data

#### Mengubah Format Tanggal dan Label Kategori

```python
# Memastikan dataset siap digunakan
print("Ukuran dataset hour:", hour_df.shape)
print("Ukuran dataset day:", day_df.shape)

print("Dataset berhasil dipertahankan tanpa penghapusan data.")
```

```python
# Mengubah kolom dteday menjadi tipe data datetime
hour_df['dteday'] = pd.to_datetime(hour_df['dteday'])
day_df['dteday'] = pd.to_datetime(day_df['dteday'])

# Memeriksa tipe data setelah perubahan
print("Tipe data dteday pada hour_df:", hour_df['dteday'].dtype)
print("Tipe data dteday pada day_df:", day_df['dteday'].dtype)
```

```python
# Mengubah kode kategori menjadi label
hour_df['season'] = hour_df['season'].map({
    1: 'Spring',
    2: 'Summer',
    3: 'Fall',
    4: 'Winter'
})

day_df['season'] = day_df['season'].map({
    1: 'Spring',
    2: 'Summer',
    3: 'Fall',
    4: 'Winter'
})

for df in [hour_df, day_df]:
    df['weathersit'] = df['weathersit'].map({
        1: 'Clear',
        2: 'Cloudy',
        3: 'Light Rain/Snow',
        4: 'Heavy Rain/Snow'
    })

hour_df['workingday'] = hour_df['workingday'].map({
    0: 'Bukan Hari Kerja',
    1: 'Hari Kerja'
})
```

```python
print(hour_df[['season', 'weathersit', 'workingday']].head())
print(day_df[['season', 'weathersit']].head())
```

**Insight:** (Opsional)
### Insight Cleaning Data

Pada tahap pembersihan data, dilakukan penyesuaian format dan kategori pada dataset `hour_df` dan `day_df` agar lebih mudah digunakan dalam proses analisis.

Pertama, ukuran kedua dataset diperiksa untuk memastikan jumlah baris dan kolom. Seluruh data dipertahankan tanpa penghapusan baris.

Kedua, kolom `dteday` pada kedua dataset diubah dari tipe data object menjadi datetime agar dapat digunakan dalam pengolahan dan analisis berdasarkan tanggal.

Ketiga, kode numerik pada kolom `season` diubah menjadi label Spring, Summer, Fall, dan Winter. Kolom `weathersit` diubah menjadi Clear, Cloudy, Light Rain/Snow, dan Heavy Rain/Snow. Khusus pada dataset `hour_df`, kolom `workingday` diubah menjadi Hari Kerja dan Bukan Hari Kerja.

Hasil pembersihan ini membuat format tanggal dan kategori data lebih mudah dipahami serta mendukung proses analisis dan visualisasi pada tahap berikutnya.

## Exploratory Data Analysis (EDA)

### Eksplorasi Rata-rata Penyewaan Berdasarkan Jenis Hari, Musim, dan Kondisi Cuaca

```python
# Mengelompokkan data berdasarkan hari kerja dan bukan hari kerja
workingday_analysis = hour_df.groupby('workingday')['cnt'].agg(
    ['mean', 'median', 'min', 'max']
).reset_index()

display(workingday_analysis)
```

```python
# Jam puncak penyewaan sepeda pada hari kerja dibandingkan dengan bukan hari kerja (2011–2012)
hourly_avg = hour_df.groupby(['workingday', 'hr'])['cnt'].mean().reset_index()

peak_hour = hourly_avg.loc[
    hourly_avg.groupby('workingday')['cnt'].idxmax()
].reset_index(drop=True)

peak_hour.columns = ['Jenis Hari', 'Jam Puncak', 'Rata-rata Penyewaan per Jam']
peak_hour['Rata-rata Penyewaan per Jam'] = peak_hour['Rata-rata Penyewaan per Jam'].round(2)

display(peak_hour)
```

```python
# Rata-rata penyewaan harian per kombinasi musim dan cuaca (2011–2012)
season_weather_summary = day_df.groupby(['season', 'weathersit'])['cnt'].agg(
    ['mean', 'count']
).reset_index()

season_weather_summary.columns = [
    'Musim', 'Kondisi Cuaca', 'Rata-rata Penyewaan Harian', 'Jumlah Hari'
]
season_weather_summary['Rata-rata Penyewaan Harian'] = (
    season_weather_summary['Rata-rata Penyewaan Harian'].round(2)
)

# Urutkan dari tertinggi ke terendah
season_weather_summary = season_weather_summary.sort_values(
    'Rata-rata Penyewaan Harian', ascending=False
).reset_index(drop=True)

display(season_weather_summary)

tertinggi = season_weather_summary.iloc[0]
terendah = season_weather_summary.iloc[-1]

print(f"Tertinggi: {tertinggi['Musim']} - {tertinggi['Kondisi Cuaca']} "
      f"({tertinggi['Rata-rata Penyewaan Harian']} per hari, {tertinggi['Jumlah Hari']} hari)")
print(f"Terendah : {terendah['Musim']} - {terendah['Kondisi Cuaca']} "
      f"({terendah['Rata-rata Penyewaan Harian']} per hari, {terendah['Jumlah Hari']} hari)")
```

```python
# Mengelompokkan rata-rata penyewaan sepeda berdasarkan musim dan kondisi cuaca
season_weather_analysis = day_df.groupby(
    ['season', 'weathersit']
)['cnt'].mean().reset_index()

# Menampilkan hasil analisis
display(season_weather_analysis)
```

**Insight:** (Opsional)
### Insight Exploratory Data Analysis (EDA)

Berdasarkan hasil analisis eksploratif, terdapat perbedaan rata-rata penyewaan sepeda antara hari kerja dan bukan hari kerja. Pengelompokan berdasarkan hari kerja memberikan gambaran awal mengenai karakteristik penyewaan pada kedua jenis hari tersebut.

Analisis berdasarkan musim dan kondisi cuaca menunjukkan bahwa rata-rata penyewaan sepeda bervariasi pada setiap kombinasi kategori. Hasil ini menjadi dasar untuk melihat lebih lanjut pola penyewaan berdasarkan waktu, musim, dan kondisi cuaca melalui visualisasi.

Penggunaan nilai rata-rata, median, minimum, dan maksimum membantu memahami distribusi penyewaan, sedangkan pengelompokan berdasarkan musim dan cuaca membantu membandingkan tingkat penyewaan antarkategori.

## Visualization & Explanatory Analysis

### Pertanyaan 1: Perbandingan Rata-rata Penyewaan Sepeda per Jam pada Hari Kerja dan Bukan Hari Kerja (2011–2012)

```python
# Perbandingan Rata-rata Penyewaan Sepeda per Jam
# antara Hari Kerja dan Bukan Hari Kerja Tahun 2011–2012

hourly_workingday = hour_df.groupby(
    ['hr', 'workingday']
)['cnt'].mean().reset_index()

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=hourly_workingday,
    x='hr',
    y='cnt',
    hue='workingday',
    marker='o'
)

plt.title(
    'Perbandingan Rata-rata Penyewaan Sepeda per Jam '
    'pada Hari Kerja dan Bukan Hari Kerja (2011–2012)'
)
plt.xlabel('Jam (00.00–23.00)')
plt.ylabel('Rata-rata Jumlah Penyewaan Sepeda')
plt.xticks(range(24))
plt.legend(title='Jenis Hari')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

### Pertanyaan 2: Rata-rata Penyewaan Sepeda Harian berdasarkan Musim dan Kondisi Cuaca (2011–2012)

```python
# Rata-rata Penyewaan Sepeda Berdasarkan Musim dan Kondisi Cuaca (2011–2012)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=season_weather_analysis,
    x='season',
    y='cnt',
    hue='weathersit',
    order=['Spring', 'Summer', 'Fall', 'Winter'],
    hue_order=['Clear', 'Cloudy', 'Light Rain/Snow']
)

plt.title(
    'Perbandingan Rata-rata Penyewaan Sepeda '
    'Berdasarkan Musim dan Kondisi Cuaca (2011–2012)'
)
plt.xlabel('Musim')
plt.ylabel('Rata-rata Jumlah Penyewaan Sepeda per Hari')
plt.legend(title='Kondisi Cuaca')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

### Interpretasi Pola Penyewaan Sepeda per Jam pada Hari Kerja dan Bukan Hari Kerja

Visualisasi garis memperlihatkan perbedaan pola rata-rata penyewaan sepeda per jam antara hari kerja dan bukan hari kerja selama tahun 2011–2012. Grafik ini memudahkan identifikasi jam-jam dengan tingkat penyewaan tinggi pada kedua jenis hari.

Pada hari kerja, pola penyewaan cenderung memiliki peningkatan pada jam berangkat dan pulang kerja. Sementara itu, pada bukan hari kerja, penyewaan cenderung lebih tinggi pada beberapa jam siang hingga sore. Perbedaan pola tersebut dapat menjadi pertimbangan dalam penyesuaian distribusi armada berdasarkan waktu.

### Interpretasi Penyewaan Sepeda Berdasarkan Musim dan Kondisi Cuaca

Berdasarkan visualisasi kedua, musim Fall dengan kondisi cuaca Clear memiliki rata-rata penyewaan tertinggi, yaitu sekitar 5.878 sepeda per hari. Sebaliknya, musim Spring dengan kondisi cuaca Light Rain/Snow memiliki rata-rata terendah, yaitu sekitar 935 sepeda per hari.

Secara umum, kondisi cuaca Clear memiliki rata-rata penyewaan lebih tinggi dibandingkan Cloudy dan Light Rain/Snow pada seluruh musim. Hal ini menunjukkan bahwa kondisi cuaca berkaitan dengan perubahan jumlah penyewaan sepeda.

Perlu diperhatikan bahwa rata-rata tersebut belum mempertimbangkan jumlah observasi pada setiap kategori cuaca.

## Analisis Lanjutan (Opsional)

```python
# Perbandingan Rata-rata Penyewaan Sepeda Berdasarkan Jam
# antara Hari Kerja dan Bukan Hari Kerja

hourly_pattern = hour_df.groupby(
    ['hr', 'workingday']
)['cnt'].mean().reset_index()

# Membuat tabel perbandingan
hourly_pivot = hourly_pattern.pivot(
    index='hr',
    columns='workingday',
    values='cnt'
)

# Mengatur urutan kolom
hourly_pivot = hourly_pivot[
    ['Bukan Hari Kerja', 'Hari Kerja']
]

display(hourly_pivot)
```

```python
# Pola Penyewaan Sepeda Berdasarkan Jam dan Kategori Hari

plt.figure(figsize=(12, 6))

plt.plot(
    hourly_pivot.index,
    hourly_pivot['Bukan Hari Kerja'],
    marker='o',
    label='Bukan Hari Kerja'
)

plt.plot(
    hourly_pivot.index,
    hourly_pivot['Hari Kerja'],
    marker='o',
    label='Hari Kerja'
)

plt.title(
    'Perbandingan Pola Penyewaan Sepeda per Jam '
    'antara Hari Kerja dan Bukan Hari Kerja (2011–2012)'
)
plt.xlabel('Jam (00.00–23.00)')
plt.ylabel('Rata-rata Jumlah Penyewaan Sepeda')
plt.xticks(range(24))
plt.legend(title='Kategori Hari')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

**Insight:** (Opsional)
Berdasarkan analisis pola penyewaan sepeda berdasarkan jam, ditemukan perbedaan pola antara hari kerja dan bukan hari kerja.

Pada hari kerja, rata-rata penyewaan tertinggi terjadi pada pukul 17.00, yaitu sebesar 525,29 penyewaan per jam. Selain itu, terjadi peningkatan penyewaan pada pagi hari, terutama pukul 07.00–08.00. Hal ini menunjukkan bahwa penggunaan sepeda pada hari kerja cenderung meningkat pada jam berangkat dan pulang aktivitas.

Sementara itu, pada bukan hari kerja, rata-rata penyewaan tertinggi terjadi pada pukul 13.00, yaitu sebesar 372,73 penyewaan per jam. Penyewaan cenderung lebih tinggi pada siang hari dibandingkan pagi hari.

Dengan demikian, pola penyewaan sepeda berbeda antara hari kerja dan bukan hari kerja. Pada hari kerja, penyewaan lebih banyak terjadi pada jam sibuk pagi dan sore, sedangkan pada bukan hari kerja penyewaan lebih terkonsentrasi pada siang hari.

Berdasarkan pola tersebut, pengelola layanan dapat meningkatkan ketersediaan sepeda pada pukul 07.00–08.00 dan 17.00 pada hari kerja. Sementara itu, pada bukan hari kerja, ketersediaan sepeda dapat lebih difokuskan pada siang hari, terutama sekitar pukul 13.00.
